# Analysis of variance (ANOVA)

## Libraries and settings

```python
# Libraries
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import scipy.stats as stats

# Settings for seaborn
sns.set_theme(style="ticks", palette="pastel")

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

# Get current working directory
print('Current working directory:', os.getcwd())
```

## Read example data (used car data)

```python
# Read data
df = pd.read_excel('t-tests_and_ANOVA.xlsx', sheet_name='anova_data')

# Show categories of 'Make'
print(df['Make'].value_counts())

# Show header
df.head()
```

## Create pivot table of mean price per make

```python
# Create pivot table
table = pd.pivot_table(df, 
                       values='Price',
                       columns=['Make'], 
                       aggfunc=np.mean)
print(table)
```

## Create grouped boxplot (groups = make, values = prices)

```python
# Plot boxplot with groups
plt.figure(figsize=(8,2))
ax = sns.boxplot(x="Price",
                 y="Make",
                 palette=['r', 'g', 'b'],
                 data=df,
                 orient='h')
plt.grid()
plt.show()
```

## Analysis of variance (ANOVA)

```python
# Create subsets (groups)
vw = df.loc[df['Make'] == 'VW']
bmw = df.loc[df['Make'] == 'BMW']
fiat = df.loc[df['Make'] == 'Fiat']

# Create ANOVA
fvalue, pvalue = stats.f_oneway(vw['Price'],
                                fiat['Price'], 
                                bmw['Price'])

# Print results
print('F-value:', fvalue.round(3), 'p-value', pvalue.round(4))
```

## Interpretation of result

<p> Because the p-value is lower than 0.05, the null hypothesis can be rejected, i.e. there is evidence that the used car prices of VW, BMW and Fiat differ. Note that, from the ANOVA, it is not clear which groups show differences of prices. For this a post-hoc test could be used (not part of the exercises).</p>

## Importing apartment data

```python
# Read the data to a pandas data frame
df = pd.read_csv('apartments_data_enriched_cleaned.csv', 
                 sep=';', 
                 encoding='utf-8')[['web-scraper-order',
                                    'address_raw',
                                    'lat',
                                    'lon',
                                    'bfs_number',
                                    'bfs_name',
                                    'rooms', 
                                    'area', 
                                    'luxurious', 
                                    'price', 
                                    'price_per_m2',
                                    'pop_dens',
                                    'frg_pct',
                                    'mean_taxable_income',
                                    'dist_supermarket']]

# Get number of rows and columns
print(df.shape)

# Show first records
df.head(5)
```

## Create pivot table with mean price_per_m2 per room size

```python
# Using pivot_table to reshape the data and calculate means 
pd.pivot_table(df[['rooms', 'price_per_m2']],
               index=['rooms'],
               values=['price_per_m2'],
               aggfunc=[np.mean, 'count'])
```

## Create grouped boxplot (groups = rooms, values = prices_per_m2)

```python
# Plot boxplot with groups
plt.figure(figsize=(8,4))
ax = sns.boxplot(x="price_per_m2",
                 y="rooms",
                 data=df,
                 orient='h')
plt.grid()
plt.show()
```

## ANOVA to compare the price_per_m2 per number of rooms (= groups)

```python
# Create subsets (groups)
r2_5 = df.loc[df['rooms'] == 2.5]['price_per_m2']
r3_5 = df.loc[df['rooms'] == 3.5]['price_per_m2']
r4_5 = df.loc[df['rooms'] == 4.5]['price_per_m2']

# Create ANOVA
fvalue, pvalue = stats.f_oneway(r2_5, r3_5, r4_5)

# Print results
print('F-value:', fvalue.round(3), 'p-value', pvalue.round(4))
```

## Interpretation of result

<p> Because the p-value is lower than 0.05, the null hypothesis can be rejected, i.e. there is evidence that the price per m2 differ between apartments with 2.5, 3.5 and 4.5 rooms. Note that, from the ANOVA, it is not clear which groups show price differences. For this a post-hoc test could be used (not part of the exercises).</p>

# ANOVA three subsets prices per m2

## New Subsets

```python
# Create subsets based on population density
low_pop_dens = df.loc[df['pop_dens'] <= 600, 'price_per_m2']
moderate_pop_dens = df.loc[(df['pop_dens'] > 600) & (df['pop_dens'] <= 1500), 'price_per_m2']
high_pop_dens = df.loc[df['pop_dens'] > 1500, 'price_per_m2']

# Print the number of entries in each subset
print('Low population density:', len(low_pop_dens))
print('Moderate population density:', len(moderate_pop_dens))
print('High population density:', len(high_pop_dens))
```

```python
# Perform one-way ANOVA
fvalue, pvalue = stats.f_oneway(low_pop_dens, moderate_pop_dens, high_pop_dens)

# Print results
print('F-value:', fvalue.round(3), 'p-value:', pvalue.round(4))

# Interpretation
alpha = 0.05
if pvalue < alpha:
    print("Reject the null hypothesis: There is evidence that the mean prices per m² differ between the groups.")
else:
    print("Fail to reject the null hypothesis: No evidence that the mean prices per m² differ between the groups.")
```

# ANOVA three subsets area

```python
# Create subsets based on population density
low_pop_dens_area = df.loc[df['pop_dens'] <= 600, 'area']
moderate_pop_dens_area = df.loc[(df['pop_dens'] > 600) & (df['pop_dens'] <= 1500), 'area']
high_pop_dens_area = df.loc[df['pop_dens'] > 1500, 'area']

# Perform one-way ANOVA
fvalue_area, pvalue_area = stats.f_oneway(low_pop_dens_area, moderate_pop_dens_area, high_pop_dens_area)

# Print results
print('F-value:', fvalue_area.round(3), 'p-value:', pvalue_area.round(4))

# Define hypotheses
print("\nNull Hypothesis: The mean areas of rental apartments are the same across the three groups of municipalities.")
print("Alternative Hypothesis: The mean areas of rental apartments differ across the three groups of municipalities.")

# Interpretation
alpha = 0.05
if pvalue_area < alpha:
    print("\nReject the null hypothesis: There is evidence that the mean areas of rental apartments differ between the groups.")
else:
    print("\nFail to reject the null hypothesis: No evidence that the mean areas of rental apartments differ between the groups.")
```

# ANOVA three subsets numbers of rooms

```python
# Perform one-way ANOVA
fvalue_rooms, pvalue_rooms = stats.f_oneway(low_pop_dens_area, moderate_pop_dens_area, high_pop_dens_area)

# Print results
print('F-value:', fvalue_rooms.round(3), 'p-value:', pvalue_rooms.round(4))

# Define hypotheses
print("\nNull Hypothesis: The mean number of rooms of rental apartments is the same across the three groups of municipalities.")
print("Alternative Hypothesis: The mean number of rooms of rental apartments differs across the three groups of municipalities.")

# Interpretation
alpha = 0.05
if pvalue_rooms < alpha:
    print("\nReject the null hypothesis: There is evidence that the mean number of rooms of rental apartments differs between the groups.")
else:
    print("\nFail to reject the null hypothesis: No evidence that the mean number of rooms of rental apartments differs between the groups.")
```
